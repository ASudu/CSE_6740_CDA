```python
import numpy as np
import scipy.io
import matplotlib.pyplot as plt
```

```python
mat = scipy.io.loadmat('sp500.mat')
mat = mat['price_move']
mat.shape
```

```python
moves = mat[:, 0]
plt.plot(moves)
plt.xlabel('Time')
plt.ylabel('Price Move')
plt.title('Price Move of S&P 500')
plt.show()
```
